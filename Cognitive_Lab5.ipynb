```python
# arr = np.array([1, 2, 3, 6, 4, 5])
# print(arr[::-1])

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

unique, count = np.unique(x, return_counts=True)
most = unique[np.argmax(count)]
print("Most frequent value:", most)
print("Indices:", np.where(x == most)[0])

unique, count = np.unique(y, return_counts=True)
most = unique[np.argmax(count)]
print("Most frequent value:", most)
print("Indices:", np.where(y == most)[0])
```

Q4

```python
pratham = np.linspace(10, 100, 25)
print("Array:", pratham)
print("Dimensions:", pratham.ndim)
print("Shape:", pratham.shape)
print("Total elements:", pratham.size)
print("Data type:", pratham.dtype)
print("Total bytes:", pratham.nbytes)
transpose = pratham.reshape(25, 1)
print("Transpose using reshape:")
print(transpose)
print("Transpose using T:")
print(pratham.T)
```

```python
ucs420_pratham = np.array([[10, 20, 30, 40],
                           [50, 60, 70, 80],
                           [90, 15, 20, 35]])
print("Array:")
print(ucs420_pratham)
print("Mean:", np.mean(ucs420_pratham))
print("Median:", np.median(ucs420_pratham))
print("Maximum:", np.max(ucs420_pratham))
print("Minimum:", np.min(ucs420_pratham))
print("Unique Elements:", np.unique(ucs420_pratham))
reshaped_ucs420_pratham = ucs420_pratham.reshape(4, 3)
print("Reshaped Array:")
print(reshaped_ucs420_pratham)
resized_ucs420_pratham = np.resize(ucs420_pratham, (2, 3))
print("Resized Array:")
print(resized_ucs420_pratham)
```
