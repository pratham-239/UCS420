Q1

```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

temperature = temperature + 2
print("Updated temperature:", temperature)

fahrenheit = (9/5) * temperature + 32
print("Temperature in Fahrenheit:", fahrenheit)

readings = temperature[temperature > 32]
print("Readings greater than 32°C:", readings)

count = np.sum(temperature > 32)
print("Number of readings greater than 32°C:", count)

# Vectorization and Boolean indexing make data processing faster by performing operations on the entire array without explicitly using a loop.
```

Q2

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

print("Total steps:", np.sum(steps))

print("Mean steps:", np.mean(steps))

print("Maximum steps:", np.max(steps))
print("Minimum steps:", np.min(steps))

print("Total steps for each day:", np.sum(steps, axis=0))

print("Total steps for each user:", np.sum(steps, axis=1))

print("Position of maximum steps:", np.unravel_index(np.argmax(steps), steps.shape))
```

Q3

```python
original = np.array([1, 2, 3, 4, 5, 6])

subset = original[1:5]
print("Original:", original)
print("Subset:", subset)

subset[0] = 999
print("Original:", original)
print("Subset:", subset)

copied = original[1:5].copy()
copied[0] = 500
print("Original:", original)
print("Copied array:", copied)

arr = np.arange(1, 13)
matrix = arr.reshape(3, 4)

print("3 x 4 Matrix:")
print(matrix)

print("First row:", matrix[0])
print("Last row:", matrix[-1])
print("Second column:", matrix[:, 1])
print("Rows 1-2 and columns 2-3:")
print(matrix[1:3, 2:4])

flat = matrix.flatten()
ravel = matrix.ravel()

print("Flatten:", flat)
print("Ravel:", ravel)

ravel[0] = 100
print("After modifying ravel:")
print("Original:", matrix)
print("Ravel:", ravel)

flat[1] = 200
print("After modifying flatten:")
print("Original:", matrix)
print("Flatten:", flat)

print("Shape:", matrix.shape)
print("Dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Data type:", matrix.dtype)
```

```python
X=np.array([
[6,70,3],
[5,50,6],
[8,80,2],
[4,30,8]
])
y=np.array([40,65,30,85])

print(f"shape: {X.shape} ndim: {X.ndim}")

X_transpose=X.T
print(X_transpose)

XtX=X.T@X
print(XtX)
XtX_inv=np.linalg.inv(XtX)
print(XtX_inv)

beta=XtX_inv@X.T@y
print(beta)

new_user=np.array([5,40,7])
print(new_user@beta)
```
