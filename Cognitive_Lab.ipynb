```python
#6.1
def addOdd(n):
  sum=0
  for i in range(1,n+1):
    if i%2 != 0:
      sum=sum+i
  return sum

n=int(input("Enter the number: "))
print(addOdd(n))

#6.2
def addPrime(n):
  sum=0
  for i in range(2,n+1):
    f=0
    for j in range(2,i):
      if i%j==0:
        f=1
        break
    if f==0:
      sum=sum+i
  return(sum)

n=int(input("Enter the number: "))
print(addPrime(n))
```
