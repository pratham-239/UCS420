Q1

```python
import pandas as pd
roll_no = 1024170190
fixed_entries = [
    {"question": "what is the annual fee", "answer": "The annual fee is Rs 500.",
     "keywords": "fee cost price charge", "category": "billing"},

    {"question": "how to reset password", "answer": "Go to Settings > Reset Password.",
     "keywords": "password reset login", "category": "account"},

    {"question": "what are your working hours", "answer": "We are open 9 AM to 5 PM.",
     "keywords": "hours timing open time", "category": "general"},

    {"question": "how can i pay the fee", "answer": "You can pay via UPI, card, or net banking.",
     "keywords": "pay payment upi fee", "category": "billing"}
]
last_digit = roll_no % 10
second_last_digit = (roll_no // 10) % 10

category = ["billing", "account", "general"]

entry1 = {
    "question": "how can I download my payment receipt",
    "answer": "Go to Payments and download your payment receipt.",
    "keywords": "payment receipt download",
    "category": category[second_last_digit % 3]
}

entry2 = {
    "question": "how can I check my pending payment",
    "answer": "Go to Payments section to check your pending payment.",
    "keywords": "pending payment status",
    "category": category[last_digit % 3]
}

fixed_entries.append(entry1)
fixed_entries.append(entry2)

df = pd.DataFrame(fixed_entries)

print(df)
```

Q2

```python
def score(query):
    query_words = query.lower().split()
    result = []
    for i in range(len(df)):
        keywords = df["keywords"][i].lower().split()
        match = 0
        for word in query_words:
            if word in keywords:
                match += 1
        confidence = match / len(query_words)
        if match > 0:
            result.append([df["question"][i], df["answer"][i], confidence])
    result.sort(key=lambda x: x[2], reverse=True)
    return result

query = input("Enter your query: ")
result = score(query)
for i in result:
    print("Question:", i[0])
    print("Answer:", i[1])
    print("Confidence:", i[2])
    print()
```

```python
def same_category(category_name, df):
    result = df[df["category"] == category_name]
    return result["question"]

category_name = "billing"
result = same_category(category_name, df)
for i in result:
    print(i)
```

```python
def score(query):
    query_words = query.lower().split()
    result = []
    for i in range(len(df)):
        keywords = df["keywords"][i].lower().split()
        match = 0
        for word in query_words:
            if word in keywords:
                match += 1
        confidence = match / len(query_words)
        if match > 0:
            result.append([df["question"][i], df["answer"][i], confidence])
    if len(result) == 0:
        print("No match found")
        return
    highest = max(i[2] for i in result)
    print("Best Matching Entries:")
    for i in result:
        if i[2] == highest:
            print("Question:", i[0])
            print("Answer:", i[1])
            print("Confidence:", i[2])
            print()
score("fee")
```
